```python
import matplotlib.pyplot as plt
from sklearn import preprocessing
from sklearn.datasets import fetch_openml
from sklearn.neural_network import MLPClassifier
import numpy as np
```

```python
# Load data
data_set= fetch_openml("mnist_784")
```

```python
# print(data_set)
# Normalize intensity of images to make it in the range [0,1] since 255 is the max (white).
# X = X / 255.0
X = preprocessing.normalize(data_set.data)
Y = data_set.target
# print(data_set.data)
```

each 2D image is now transformed into a 1D vector with dimension [1, 28x28] = [1, 784]


```python
print(X.shape)
```

```python
print(Y.shape)
```

```python
for i in range(6):
  plt.subplot(2,3,i + 1)
  plt.imshow(X[i].reshape(28, 28), cmap=plt.get_cmap("gray"))
  plt.title(f'This number is: {Y[i]}')
plt.show()
```

```python
# Split the data into train/test sets
X_train, X_test = X[:60000], X[60000:]
Y_train, Y_test = Y[:60000], Y[60000:]
# print(list(Y_test))
```

```python
classifiers = list()
solvers = ['sgd','adam', 'lbfgs']
activations = ['relu', 'identity', 'logistic', 'tanh']
```

```python
temp_classifier = list()
for j in range(len(activations)):
  classifier = MLPClassifier(
      hidden_layer_sizes = (50,20,10),
      activation = activations[j],  # Activation function for the hidden layer.
      solver = solvers[0], #sgd
      alpha=1e-4,
      learning_rate = 'adaptive', # if training loss not keeps decreasing then the current learning rate is divided by 5.
      learning_rate_init = 0.1,
      max_iter = 100, # the number of epochs (how many times each data point will be used), not the number of gradient steps.
      shuffle = True,
      random_state = 1,
      verbose = 10
      )
    # fit the model on the training data
  classifier.fit(X_train, Y_train)
  temp_classifier.append(classifier)
classifiers.append(temp_classifier)
```

```python
# Solver = Adam
temp_classifier = list()
for j in range(len(activations)):
  classifier = MLPClassifier(
      hidden_layer_sizes=(50,20,10),
      activation = activations[j],  # Activation function for the hidden layer.
      solver= solvers[1], #adams
      alpha=1e-4,
      learning_rate_init=0.1,
      max_iter = 100, # the number of epochs (how many times each data point will be used), not the number of gradient steps.
      shuffle = True,
      random_state=1,
      verbose=10
      )
    # fit the model on the training data
  classifier.fit(X_train, Y_train)
  temp_classifier.append(classifier)
classifiers.append(temp_classifier)
```

```python
predictions = list()
for i in range(len(solvers) - 1):
  temp_prediction = list()
  for j in range(len(activations)):
    predicted_y= classifiers[i][j].predict(X_test)
    temp_prediction.append(predicted_y)
  predictions.append(temp_prediction)
```

```python
for i in range(2):
  for j in range(4):
    plt.subplot(2,4,4 * i + j + 1)
    plt.imshow(X_test[i].reshape(28, 28), cmap=plt.get_cmap("gray"))
    plt.title(f'{list(Y_test)[2*i + j]} is pred: {predictions[i][j][2*i + j]}')
plt.show()
```

```python

# import module
from tabulate import tabulate
test_score = list()
for i in range(len(solvers) - 1):
  temp_score = list()
  for j in range(len(activations)):
    temp_score.append(classifiers[i][j].score(X_test, Y_test))
  test_score.append(temp_score)

# assign data
mydata = [["Sgd",test_score[0][0], test_score[0][1], test_score[0][2], test_score[0][3]],
    ["Adam",test_score[1][0], test_score[1][1], test_score[1][2], test_score[1][3]]]

# create header

head = ["Solver \ Activation",'relu', 'identity', 'logistic', 'tanh']
# display table
print("Test Score")
print(tabulate(mydata, headers=head, tablefmt="grid"))
```

```python
train_score = list()
for i in range(len(solvers) - 1):
  temp_score = list()
  for j in range(len(activations)):
    temp_score.append(classifiers[i][j].score(X_test, Y_test))
  train_score.append(temp_score)

# assign data
mydata = [
    ["Sgd",train_score[0][0],train_score[0][1], train_score[0][2], train_score[0][3]],
    ["Adam",train_score[1][0], train_score[1][1], train_score[1][2], train_score[1][3]]
]

# create header
head = ["Solver \ Activation",'relu', 'identity', 'logistic', 'tanh']
# display table
print("Train Score")
print(tabulate(mydata, headers=head, tablefmt="grid"))
```

```python
fig, axes = plt.subplots(1, 1, figsize=(10,10))
colors = [['blue', 'orange', 'green', 'red', 'purple'],['brown', 'pink', 'gray', 'olive', 'cyan']]
for i in range(2):
  for j in range(4):
    axes.plot(classifiers[i][j].loss_curve_, '-', color = colors[i][j])
axes.set_xlabel("number of iteration")
axes.set_ylabel("loss")
plt.show()
```
